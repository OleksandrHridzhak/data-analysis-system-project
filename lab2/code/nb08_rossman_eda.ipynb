# Rossmann Store Sales - Exploratory Data Analysis (nb08)
Reference EDA notebook exploring the 100-store dataset.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

train_df = pd.read_csv('../data/rossm_train_df.csv', dtype={'StateHoliday': str})
test_df = pd.read_csv('../data/rossm_test_df.csv', dtype={'StateHoliday': str})
print('Train shape:', train_df.shape)
print('Test shape:', test_df.shape)
train_df.head()
```

## Data Overview & Distributions


```python
print('Stores count:', train_df['Store'].nunique())
print('Date range train:', train_df['Date'].min(), 'to', train_df['Date'].max())
print('Date range test:', test_df['Date'].min(), 'to', test_df['Date'].max())
print('Closed pct:', (train_df['Open'] == 0).mean() * 100)
```

```python
# Sales & logSales distribution on open days
open_train = train_df[train_df['Open'] == 1]
print('Sales descriptive stats:')
print(open_train['Sales'].describe())
```

## Sales vs Promo and Weekday


```python
# Boxplot of Sales by Weekday split by Promo
promo_weekday = open_train.groupby(['weekday', 'Promo'])['Sales'].mean().unstack()
print(promo_weekday)
```

## Sales vs Customers and CompetitionDistance


```python
corr = open_train[['Sales', 'Customers', 'CompetitionDistance', 'logSales']].corr()
print('Correlation matrix:')
print(corr)
```
